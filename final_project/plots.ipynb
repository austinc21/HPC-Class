```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams.update({"font.size": 12})


N = [10, 100, 1000]
cores_64 = [120.761179, 183.939373, 854.716497]
cores_8 = [146.200324, 638.038990, np.nan]
cores_0 = [88.5747, 817.267, 8036.31]

ls = ['--']
# plt.loglog(N, cores_64, label='64 cores', linestyle=ls[0], color = 'red')
# plt.loglog(N, cores_8, label='8 cores', linestyle=ls[0], color = 'blue')
plt.loglog(N, cores_0, label='Non-Parallel', linestyle=ls[0], color = 'green')
plt.legend()
plt.xlabel(r'$\log{N}$')
plt.ylabel('Time (s)')
plt.grid()
plt.tight_layout()
plt.show()
```

```python
# Worker Comparison
cores = [1, 2, 4, 8, 16, 32, 64]
cores_time = [600.89, 374.28, 175.60, 105.43, 55.63, 33.49, 21.05]
plt.plot(cores, cores_time, linestyle='--', color='blue')
plt.xticks(cores, cores)
plt.xscale('log', base=2)
plt.grid()
plt.xlabel('Number of Cores')
plt.ylabel('Time (s)')
# plt.title('Worker Performance Comparison, N = 100')
plt.show()

plt.plot(cores, cores_time, linestyle='--', color='blue')
plt.xticks(cores, cores)
plt.xscale('log', base=2)
plt.yscale('log')
plt.yticks([1e1, 1e2, 1e3], [r'$10^1$', r'$10^2$', r'$10^3$'])
plt.grid()
plt.xlabel('Number of Cores')
plt.ylabel('Time (s)')
# plt.title('Worker Performance Comparison, N = 100')
plt.show()
```

```python
# Serial vs 64 core parallel
N  = [10, 100, 1000]
serial_time = [102.791, 1019.14, 10113]
parallel_64_time = [120.761179, 183.939373, 854.716497]

plt.loglog(N, serial_time, label='Serial (NumPy)', linestyle='--', color='red')
plt.loglog(N, parallel_64_time, label='Parallel (64 cores)', linestyle='-', color='blue')
plt.xlabel('N')
plt.ylabel('Time (s)')
plt.grid()
plt.legend()
plt.show()
```
